# 정형 데이터와 비정형 데이터
**정형 데이터(structured data)**: 어떠한 구조로 일정하게 이루어져 있는 데이터, 특정 구조로 정리가 가능함.

**비정형 데이터(unstructured data)**: 어떠한 구조로 되어 있지 않고 특정 구조로 정리하기 어려운 것들\
예: 책과 글과 같은 텍스트 데이터, 음성 데이터, 사진 등

정형 데이터를 다루는 데 가장 뛰어난 알고리즘이 **앙상블 학습(ensemble learning)** 이다.\
앙상블 학습은 대부분 결정 트리를 기반으로 만들어져 있다.

비정형 데이터의 경우에는 딥러닝을 이용해야 한다. 신경망 알고리즘의 발전에 따라 이를 이용해 비정형 데이터를 학습시켜 여러 모델을 제작할 수 있게 되었다.

## 랜덤 포레스트
**랜덤 포레스트(Random Forest)**는 앙상블 학습(Ensemble Learning)의 대표적인 모델 중 하나이다.\
안정적이고 뛰어난 성능을 보여 실제 머신러닝 문제에서 매우 널리 사용된다.

랜덤 포레스트는 **여러 개의 결정 트리(Decision Tree)를 서로 독립적으로 학습**시킨 뒤, 각각의 예측 결과를 종합하여 최종 예측을 수행한다.\
여기서 각각의 결정 트리를 나무(Tree), 여러 개의 나무를 모아 만든 모델을 숲(Forest)에 비유하여 **랜덤 포레스트(Random Forest)**라는 이름이 붙었다.


### 랜덤(Random)의 의미
랜덤 포레스트에서 **랜덤**이라는 말은 다음 두 가지 과정에서 사용된다.

1. 훈련 데이터를 랜덤하게 선택한다.
2. 특성(Feature)을 랜덤하게 선택한다.

### 부트스트랩(Bootstrap)
랜덤 포레스트는 모델을 훈련하기 위해 입력된 훈련 데이터에서 랜덤하게 샘플을 추출하여 각각의 결정 트리마다 별도의 훈련 데이터를 생성한다.\
이 과정에서는 **복원 추출(Sampling with Replacement)**을 사용한다. 즉, 한 번 선택된 샘플도 다시 선택될 수 있기 때문에 동일한 데이터가 여러 번 포함될 수 있다.\
이러한 방법을 **부트스트랩(Bootstrap)**이라고 하며, 만들어진 데이터를 **부트스트랩 샘플(Bootstrap Sample)**이라고 한다.

> 기본적으로 하나의 부트스트랩 샘플의 크기는 원본 훈련 데이터와 동일하다.
>
> 예를 들어 훈련 데이터가 100개라면 부트스트랩 샘플도 100개를 추출하지만, 중복 추출이 가능하므로 일부 데이터는 여러 번 선택되고 일부 데이터는 한 번도 선택되지 않을 수 있다.

예시)

```text
원본 데이터
A B C D E

부트스트랩 샘플
A C C E B
```

위 예시에서 C는 두 번 선택되었고 D는 선택되지 않았다.

### 특성의 랜덤 선택
랜덤 포레스트는 데이터를 랜덤하게 선택하는 것뿐만 아니라 **각 노드에서 사용할 특성도 랜덤하게 선택**한다.\
예를 들어 특성이 다음과 같이 있다고 하자.

```text
키
몸무게
나이
성별
```

결정 트리는 매번 네 개의 특성을 모두 고려하지만,
랜덤 포레스트는 노드를 생성할 때마다

```text
몸무게, 나이
```

또는

```text
키, 성별
```

처럼 일부 특성만 무작위로 선택하여 분할 기준을 찾는다.\
이 덕분에 여러 결정 트리가 서로 다른 구조를 가지게 되어 과적합을 줄일 수 있다.

### 분류 모델
분류 모델인 `RandomForestClassifier`는 기본적으로 **전체 특성 개수의 제곱근(`sqrt`)만큼의 특성**을 선택하여 각 노드를 분할한다.\
예를 들어 특성이 16개라면

```text
√16 = 4
```

즉, 노드마다 4개의 특성만 무작위로 선택하여 최적의 분할을 찾는다.


### 회귀 모델
회귀 모델인 `RandomForestRegressor`는 기본적으로 **전체 특성을 모두 사용(`max_features=1.0`)**하여 분할을 수행한다.\
필요에 따라 `max_features` 매개변수를 변경하여 사용할 특성의 개수를 제한할 수도 있다.


### 최종 예측 방법
랜덤 포레스트는 여러 결정 트리의 예측을 하나로 합쳐 최종 결과를 만든다.

* **분류(Classification)**: 가장 많은 트리가 선택한 클래스를 최종 결과로 사용한다. (다수결 투표, Majority Voting)
* **회귀(Regression)**: 모든 트리의 예측값의 평균을 최종 결과로 사용한다.

예를 들어 분류 문제에서 5개의 트리가 다음과 같이 예측하였다면,

```text
트리1 → 고양이
트리2 → 강아지
트리3 → 고양이
트리4 → 고양이
트리5 → 강아지
```

최종 결과는 **고양이**가 된다.

회귀 문제에서는

```text
120
130
125
122
128
```

과 같이 예측했다면 평균값인

```text
125
```

를 최종 예측값으로 사용한다.

### 랜덤 포레스트의 장점
* 높은 예측 성능을 보이는 경우가 많다.
* 과적합이 결정 트리보다 훨씬 적다.
* 데이터의 스케일링(StandardScaler)이 거의 필요 없다.
* 특성 중요도(Feature Importance)를 확인할 수 있다.
* 결측치와 이상치에 비교적 강하다.

### 랜덤 포레스트의 단점
* 하나의 결정 트리보다 학습 시간이 오래 걸린다.
* 많은 트리를 사용할수록 메모리 사용량이 증가한다.
* 모델의 구조를 사람이 해석하기 어렵다.
* 딥러닝이나 부스팅 계열(XGBoost, LightGBM)보다 성능이 낮은 경우도 있다.

### 결정 트리와의 차이
| 결정 트리       | 랜덤 포레스트      |
| ----------- | ------------ |
| 하나의 트리만 사용  | 여러 개의 트리를 사용 |
| 과적합되기 쉬움    | 과적합이 크게 감소   |
| 학습 속도가 빠름   | 학습 속도가 더 느림  |
| 결과를 해석하기 쉬움 | 해석이 어려움      |
| 예측 성능이 불안정  | 예측 성능이 안정적   |
