# 앙상블 학습

**앙상블 학습(Ensemble Learning)** 은 여러 개의 학습 모델(기본 학습기, base learner)을 학습시킨 뒤 각 모델의 예측을 결합하여 하나의 최종 예측을 만드는 기계학습 기법이다. 하나의 모델에만 의존하지 않고 여러 모델의 판단을 모아 예측의 정확도와 안정성을 높이는 것이 목적이다.

모델마다 학습 데이터의 서로 다른 특징을 학습하고, 잘못 예측하는 사례도 다를 수 있다. 각 모델의 예측 오류가 서로 다르면 한 모델의 실수를 다른 모델의 올바른 예측이 보완할 수 있다. 따라서 앙상블의 효과는 단순히 모델 수를 늘리는 것보다 서로 다른 특성을 가진 모델들의 예측을 적절히 결합하는 데 달려 있다.

분류 문제에서는 여러 모델의 예측 중 가장 많은 표를 얻은 결과를 선택하는 **다수결 투표(hard voting)** 나, 각 모델의 클래스별 확률을 평균 또는 가중 평균하여 결정하는 **소프트 투표(soft voting)** 를 사용할 수 있다. 회귀 문제에서는 각 모델의 예측값을 평균하거나 가중 평균하여 최종값을 구하는 방식이 일반적이다.

앙상블은 개별 모델보다 성능을 높이고 예측을 안정화할 수 있지만, 항상 성능 향상을 보장하지는 않는다. 모델들이 비슷한 오류를 내면 결합 효과가 작을 수 있고, 여러 모델을 학습·관리하는 계산 비용이 커지거나 최종 예측을 설명하기 어려워질 수 있다. 대표적인 앙상블 방법으로 배깅(Bagging), 부스팅(Boosting), 스태킹(Stacking)이 있다.

이 문서에서 알아볼 앙상블 학습 모델들은 4가지이다.
- 랜덤 포레스트(Random Forest): 여러 개의 결정 트리(Decison Tree)를 결합하여 만든 앙상블 모델.
- 엑스트라 트리(Extra Tree): 랜덤 포레스트아 비슷하게 여러 개의 결정 트리를 이용하지만, 각각의 트리의 노드를 분할하는 과정을 랜덤하게 수행한다.
 - 그레이디언트 부스팅(gradient boosting): `depth`가 얕은 결정 트리들을 이용하여 이전 트리들의 오차를 보완하는 기법을 이용한다.

# 정형 데이터와 비정형 데이터
**정형 데이터(structured data)**: 어떠한 구조로 일정하게 이루어져 있는 데이터, 특정 구조로 정리가 가능함.

**비정형 데이터(unstructured data)**: 어떠한 구조로 되어 있지 않고 특정 구조로 정리하기 어려운 것들\
예: 책과 글과 같은 텍스트 데이터, 음성 데이터, 사진 등

정형 데이터를 다루는 데 가장 뛰어난 알고리즘이 **앙상블 학습(ensemble learning)** 이다.\
앙상블 학습은 대부분 결정 트리를 기반으로 만들어져 있다.

비정형 데이터의 경우에는 딥러닝을 이용해야 한다. 신경망 알고리즘의 발전에 따라 이를 이용해 비정형 데이터를 학습시켜 여러 모델을 제작할 수 있게 되었다.

## 랜덤 포레스트
**랜덤 포레스트(Random Forest)**는 앙상블 학습(Ensemble Learning)의 대표적인 모델 중 하나이다.\
안정적이고 뛰어난 성능을 보여 실제 머신러닝 문제에서 매우 널리 사용된다.

랜덤 포레스트는 **여러 개의 결정 트리(Decision Tree)를 서로 독립적으로 학습**시킨 뒤, 각각의 예측 결과를 종합하여 최종 예측을 수행한다.\
여기서 각각의 결정 트리를 나무(Tree), 여러 개의 나무를 모아 만든 모델을 숲(Forest)에 비유하여 **랜덤 포레스트(Random Forest)**라는 이름이 붙었다.


### 랜덤(Random)의 의미
랜덤 포레스트에서 **랜덤**이라는 말은 다음 두 가지 과정에서 사용된다.

1. 훈련 데이터를 랜덤하게 선택한다.
2. 특성(Feature)을 랜덤하게 선택한다.

### 부트스트랩(Bootstrap)
랜덤 포레스트는 모델을 훈련하기 위해 입력된 훈련 데이터에서 랜덤하게 샘플을 추출하여 각각의 결정 트리마다 별도의 훈련 데이터를 생성한다.\
이 과정에서는 **복원 추출(Sampling with Replacement)**을 사용한다. 즉, 한 번 선택된 샘플도 다시 선택될 수 있기 때문에 동일한 데이터가 여러 번 포함될 수 있다.\
이러한 방법을 **부트스트랩(Bootstrap)**이라고 하며, 만들어진 데이터를 **부트스트랩 샘플(Bootstrap Sample)**이라고 한다.

> 기본적으로 하나의 부트스트랩 샘플의 크기는 원본 훈련 데이터와 동일하다.
>
> 예를 들어 훈련 데이터가 100개라면 부트스트랩 샘플도 100개를 추출하지만, 중복 추출이 가능하므로 일부 데이터는 여러 번 선택되고 일부 데이터는 한 번도 선택되지 않을 수 있다.

예시)

```text
원본 데이터
A B C D E

부트스트랩 샘플
A C C E B
```

위 예시에서 C는 두 번 선택되었고 D는 선택되지 않았다.

### 특성의 랜덤 선택
랜덤 포레스트는 데이터를 랜덤하게 선택하는 것뿐만 아니라 **각 노드에서 사용할 특성도 랜덤하게 선택**한다.\
예를 들어 특성이 다음과 같이 있다고 하자.

```text
키
몸무게
나이
성별
```

결정 트리는 매번 네 개의 특성을 모두 고려하지만,
랜덤 포레스트는 노드를 생성할 때마다

```text
몸무게, 나이
```

또는

```text
키, 성별
```

처럼 일부 특성만 무작위로 선택하여 분할 기준을 찾는다.\
이 덕분에 여러 결정 트리가 서로 다른 구조를 가지게 되어 과적합을 줄일 수 있다.

### 분류 모델
분류 모델인 `RandomForestClassifier`는 기본적으로 **전체 특성 개수의 제곱근(`sqrt`)만큼의 특성**을 선택하여 각 노드를 분할한다.\
예를 들어 특성이 16개라면

```text
√16 = 4
```

즉, 노드마다 4개의 특성만 무작위로 선택하여 최적의 분할을 찾는다.


### 회귀 모델
회귀 모델인 `RandomForestRegressor`는 기본적으로 **전체 특성을 모두 사용(`max_features=1.0`)**하여 분할을 수행한다.\
필요에 따라 `max_features` 매개변수를 변경하여 사용할 특성의 개수를 제한할 수도 있다.


### 최종 예측 방법
랜덤 포레스트는 여러 결정 트리의 예측을 하나로 합쳐 최종 결과를 만든다.

* **분류(Classification)**: 가장 많은 트리가 선택한 클래스를 최종 결과로 사용한다. (다수결 투표, Majority Voting)
* **회귀(Regression)**: 모든 트리의 예측값의 평균을 최종 결과로 사용한다.

예를 들어 분류 문제에서 5개의 트리가 다음과 같이 예측하였다면,

```text
트리1 → 고양이
트리2 → 강아지
트리3 → 고양이
트리4 → 고양이
트리5 → 강아지
```

최종 결과는 **고양이**가 된다.

회귀 문제에서는

```text
120
130
125
122
128
```

과 같이 예측했다면 평균값인

```text
125
```

를 최종 예측값으로 사용한다.

### 랜덤 포레스트의 장점
* 높은 예측 성능을 보이는 경우가 많다.
* 과적합이 결정 트리보다 훨씬 적다.
* 데이터의 스케일링(StandardScaler)이 거의 필요 없다.
* 특성 중요도(Feature Importance)를 확인할 수 있다.
* 결측치와 이상치에 비교적 강하다.

### 랜덤 포레스트의 단점
* 하나의 결정 트리보다 학습 시간이 오래 걸린다.
* 많은 트리를 사용할수록 메모리 사용량이 증가한다.
* 모델의 구조를 사람이 해석하기 어렵다.
* 딥러닝이나 부스팅 계열(XGBoost, LightGBM)보다 성능이 낮은 경우도 있다.

### 결정 트리와의 차이
| 결정 트리       | 랜덤 포레스트      |
| ----------- | ------------ |
| 하나의 트리만 사용  | 여러 개의 트리를 사용 |
| 과적합되기 쉬움    | 과적합이 크게 감소   |
| 학습 속도가 빠름   | 학습 속도가 더 느림  |
| 결과를 해석하기 쉬움 | 해석이 어려움      |
| 예측 성능이 불안정  | 예측 성능이 안정적   |


`RamdomForestClassifier` 클래스를 이용해 이전에 사용하였던 데이터셋을 가져오는 문제에 적용해보자.

````python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

wine = pd.read_csv("https://raw.githubusercontent.com/rickiepark/hg-mldl/master/wine.csv")
data = wine[['alcohol', 'sugar', 'pH']]
target = wine['class']
train_input, test_input, train_target, test_target = train_test_split(data, target, test_size=0.2, random_state=42)
````

교차 검증까지 함께 사용한다.
>`RandomForestClassifier`는 기본적으로 100개의 `DecisionTree` 개체를 생성하므로, 모든 cpu를 이용할 수 있도록 한다.(`n_jobs=-1`)
> `cross_validate()`또한 모든 cpu를 사용한다.

````python
from sklearn.model_selection import cross_validate
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_jobs=-1, random_state=42)
scores = cross_validate(rf, train_input, train_target, return_train_score=True, n_jobs=-1)

# print(scores)
print(f"학습세트 점수: {np.mean(scores['train_score'])}, 테스트 세트 점수: {np.mean(scores['test_score'])}")
````

여기까지 `RandomForestClasifier`를 사용해보았다.
>모델은 훈련 세트에 다소 과적합 되어있다.

`RandomForestClassifier`는 `DecisionTreeclassifier`의 앙상블이기 때문에, `DecisonTreeClassifer`가 제공하는 대부분의 주요 매개변수를 모두 제공한다.
또한, `DecisonTreeClassifier`에서 사용 가능한 특성 중요도 파악도 가능하다.
>`RandomForestClassifier`의 특성 중요도는 각각의 결정 트리의 특성 중요도를 취합한 것이다.
> 특성 중요도는 `feature_importances_`에 저장된다.

````python
rf.fit(train_input, train_target)
print(rf.feature_importances_)
````

이전(하나의 `DecisionTreeClassifier`만 사용)의 특성 중요도는 다음과 같았다.
```
Alc, sugar, pH
특성 중요도: [0.12345626 0.86862934 0.0079144 ]
```
`sugar`의 특성 중요도가 가장 높았다.\
현재의 특성 중요도는:
```
Alc, sugar, pH
[0.23167441 0.50039841 0.26792718]
```
위와 같다.

특성 중요도가 일부 변한 것을 확인할 수 있다.

또한, `RandomForestClasifier`에는 자체적으로 모델을 평가하는 점수가 있다.\
데이터 세트에서 랜덤하게 추출한 부트스트랩 샘플을 만들어 훈련을 진행하는데, 이 과정에서 포함되지 않고 남는 샘플(OOB, out of bag)을 이용하여 평가하는 데에 사용한다.

````python
rf_oob = RandomForestClassifier(oob_score=True, n_jobs=-1, random_state=42)
rf_oob.fit(train_input, train_target)
print(rf_oob.oob_score_)
````

## 엑스트라 트리
**엑스트라 트리(Extra Tree)** 는 Random Forest와 매우 유사하게 작동한다.

다만 엑스트라 트리는 부트스트랩 샘플을 이용하지 않는다. 결정 트리를 만들 때 모든 훈련 세트를 이용하고,
**노드를 분할할 때 가장 좋은 분할을 찾는 것이 아니라 랜덤으로 분할한다.**
> 지니 불순도나 엔트로피 불순도 등을 고려하지 않는다.
> `DecisionTreeClassifier`의 `splitter`를 `random`으로 지정한 것과 같다.

### 기대 효과:
하나의 결정 트리에서 노드를 무작위로 분할한다면, 모델 자체의 성능은 낮아지겠지만,\
그만큼 한 번에 많은 트리를 결합하여 모델을 만들기 때문에 테스트 세트에 의존하는 과대적합을 막을 수 있다.

sklearn에서 엑스트라 트리는 `ExtraTreesClassifier`이다.

````python
from sklearn.ensemble import ExtraTreesClassifier
et = ExtraTreesClassifier(n_jobs=-1, random_state=42)
scores = cross_validate(et, train_input, train_target, return_train_score=True, n_jobs=-1)

print(f"학습 세트 점수: {np.mean(scores['train_score'])}, 테스트 세트 점수: {np.mean(scores['test_score'])}")
````

> 비슷하게 과대적합되었지만, 이는 학습 세트의 양이 적기 때문이다.

일반적으로 엑스트라 트리의 무작위성으로 인해 랜덤 포레스트보다 더 많은 데이터를 훈련하는 것이 좋다.\
데이터의 훈련 량은 더 많이 요구하지만, 학습의 속도는 랜덤 포레스트에 비해 랜덤으로 노드를 분할하므로 더 빠르다.
> 일반적은 결정 트리에서는 분할의 최적 값을 계산하는데에 시간이 많이 소요된다.
> 하지만 엑스트라 트리에서는 노드를 분할하는 과정을 랜덤으로 수행하므로 학습 시간을 줄인다.

특성 중요도를 확인해보자.

````python
et.fit(train_input, train_target)
print(et.feature_importances_)
````

마찬가지로 `sugar`의 중요도를 가장 높이 평가하였다.

## 그레이디언트 부스팅
그레이디언트 부스팅(gradient boosting)은 깊이가 얕은 결정 트리들을 이용하여 이전 트리의 오차를 줄여나가는 방식으로 모델을 앙상블한다.

`sklearn`에서는 그레이디언트 부스팅 모델을 `GradientBoostingClassifier`가 제공한다.
> `GradientBoostingClassifier`의 최대 깊이 기본값은 `3`이다.
>즉, 깊이가 3인 결정 트리를 이용한다.

또한, 경사 하강법을 이용하여 트리를 앙상블에 추가한다.
### 학습 원리

그레이디언트 부스팅은 여러 트리를 **순차적으로** 학습한다. 첫 번째 트리가 예측한 뒤, 다음 트리는 현재 모델이 잘못 예측한 부분을 보완하도록 학습한다. 이 과정을 반복하면서 각 트리의 예측을 더해 최종 예측을 만든다.

회귀에서는 흔히 이전 예측과 실제값의 차이를 다음 트리가 학습한다. 분류에서는 손실 함수의 **음의 그레이디언트(의사 잔차)** 를 이용해 현재 모델의 오차를 줄이는 방향으로 트리를 추가한다. 따라서 각 트리는 독립적으로 학습되는 랜덤 포레스트와 달리, 앞서 학습된 트리의 결과에 영향을 받는다.

### 주요 매개변수

- `n_estimators`: 앙상블에 추가할 트리의 수이다. 트리가 많아지면 학습 시간이 늘고, 지나치게 많으면 과적합될 수 있다.
- `learning_rate`: 각 트리가 최종 예측에 기여하는 크기를 조절한다. 값을 낮추면 보통 더 많은 트리가 필요하지만, 학습을 더 신중하게 진행할 수 있다.
- `max_depth`: 개별 트리의 깊이를 제한한다. 얕은 트리를 사용하면 각 단계에서 복잡한 패턴을 한꺼번에 학습하는 것을 억제한다.

일반적으로 `learning_rate`와 `n_estimators`는 함께 조정한다. 성능은 데이터와 매개변수에 따라 달라지므로 교차 검증으로 적절한 조합을 찾는 것이 좋다.

````python
from sklearn.ensemble import GradientBoostingClassifier
gb = GradientBoostingClassifier(random_state=42)
scores = cross_validate(gb, train_input, train_target, return_train_score=True, n_jobs=-1)
print(np.mean(scores['train_score']), np.mean(scores['test_score']))
````

거의 과대적합이 되지 않은 모습을 볼 수 있다.

그레이디언트 부스팅은 결정 트리의 개수를 늘여도 과대적합이 쉽게 되지 않는다.

학습률을 증가시키고 트리의 개수를 늘여보겠다.

````python
gb_high = GradientBoostingClassifier(n_estimators=500, learning_rate=0.2, random_state=42)
scores_high = cross_validate(gb_high, train_input, train_target, return_train_score=True, n_jobs=-1)
print(np.mean(scores_high['train_score']), np.mean(scores_high['test_score']))
````

결정 트리 개수를 기본값에 비해 500개로 5배를 늘이고, 학습률을 2배하였다.

````python
gb_high.fit(train_input, train_target)
print(gb_high.feature_importances_)
````

특성 중요도 또한 `sugar`을 가장 중요하다고 평가하였다.

>`subsample`이 1보다 작다면 그 비율만큼 훈련 세트를 이용한다.

일반적으로 그레이디언트 부스팅이 랜덤 포레스트보다 더 높은 성능을 보인다.\
다만, 트리를 추가하는 순서 또한 계산되어야 하므로 훈련 속도가 비교적 느리다.
>`GradientBoostingClassifier`에는 `n_jobs` 매개변수가 존재하지 않는다.

## 히스토르그램 기반 그레이디언트 부스팅(Historgram-based Gradient Boosting)
히스토그램 기반 그레이디언트 부스팅은 정형 데이터 머신러닝 알고리즘 중 가장 많이 쓰인다.

## 히스토르그램 기반 그레이디언트 부스팅(Histogram-based Gradient Boosting)

히스토그램 기반 그레이디언트 부스팅은 정형 데이터 머신러닝 알고리즘 중 가장 많이 쓰인다.

기존의 `GradientBoostingClassifier`는 각 분할 지점마다 연속적인 값을 비교하며 학습해야 하므로 계산 비용이 크다. 반면 히스토그램 기반 그레이디언트 부스팅은 특성 값을 구간(bin)으로 나누어 히스토그램을 만들고, 이 히스토그램을 기반으로 최적의 분할을 찾는다. 이렇게 하면 학습 속도가 빠르고 메모리 사용량도 더 효율적이다.

핵심 아이디어는 다음과 같다.

- 특성 값을 연속값 그대로 다루는 대신 여러 구간으로 나눈다.
- 각 구간별로 누적된 정보를 이용해 분할 후보를 빠르게 계산한다.
- 얕은 트리 여러 개를 순차적으로 학습하면서 이전 트리의 오차를 보완한다.
- 누락값을 처리하는 기능도 기본적으로 제공한다.

이 방식은 특히 정형 데이터에서 매우 강력한 성능을 보여준다. 일반적으로 `RandomForest`보다 높은 예측 성능을 내는 경우가 많고, `GradientBoosting`보다 훨씬 빠르게 학습할 수 있다는 장점이 있다.

`sklearn`에서는 `HistGradientBoostingClassifier`와 `HistGradientBoostingRegressor`를 제공한다. 이 모델은 보통 정규화나 스케일링이 필수적이지는 않지만, 특성 값의 분포가 매우 skewed 되어 있거나 이상치가 많을 때는 전처리를 적절히 고려할 수 있다.

또한 히스토그램 기반 부스팅은 각 트리가 이전 트리의 오차를 줄이는 방향으로 학습되므로, 깊이가 얕은 결정 트리의 조합으로도 좋은 성능을 낼 수 있다. 이 점 때문에 적은 깊이의 트리를 여러 개 연결하는 앙상블 방식에 매우 적합하다.

실무적으로는 대용량 정형 데이터에서 특히 강력한 성능을 보인다. 속도와 성능의 균형이 좋고, 대규모 데이터셋에서도 비교적 안정적으로 동작한다는 점에서 많은 데이터 사이언티스트와 머신러닝 엔지니어가 선호하는 모델 중 하나이다.

예를 들어, `HistGradientBoostingClassifier`는 다음과 같은 주요 매개변수를 가진다.

- `max_depth`: 각 트리의 최대 깊이
- `learning_rate`: 각 트리가 보완하는 정도
- `max_iter`: 트리의 총 개수
- `max_leaf_nodes`: 리프 노드의 최대 개수
- `l2_regularization`: 가중치 규제 강도

이 매개변수들을 조절하면서 교차 검증을 통해 적절한 모델을 찾는 것이 일반적이다.

결론적으로, 히스토그램 기반 그레이디언트 부스팅은 정형 데이터에서 가장 많이 쓰이는 앙상블 알고리즘 중 하나이며, 빠른 학습 속도와 높은 예측 성능을 동시에 제공한다는 점에서 매우 중요하다.

````python
from sklearn.ensemble import HistGradientBoostingClassifier
hgb = HistGradientBoostingClassifier(random_state=42)
scores_hgb = cross_validate(hgb, train_input, train_target, return_train_score=True, n_jobs=-1)
print(np.mean(scores_hgb['train_score']), np.mean(scores_hgb['test_score']))
````

그레이디언트 부스팅보다 더 높은 성능을 보인다.

특성 중요도를 확인하기 위해서는, `HistGradientBoostingClassifier`에는 내부적으로 특성 중요도를 제공하는 속성이 존재하지 않으므로 외부 함수인 `permutation_importance()`를 이용하여야 한다.

`permutation_importance()`는 각 특성을 무작위로 섞었을 때 모델의 성능이 얼마나 떨어지는지를 측정한다. 값이 클수록 해당 특성이 예측에 큰 영향을 준다는 뜻이며, 히스토그램 기반 그레이디언트 부스팅처럼 `feature_importances_`를 제공하지 않는 모델에서도 유용하게 사용할 수 있다.

````python
from sklearn.inspection import permutation_importance

hgb.fit(train_input, train_target)

perm_importance = permutation_importance(
    hgb, train_input, train_target,
    n_repeats=10, random_state=42, n_jobs=-1 # n_repeats 매개변수는 모델을 랜덤하게 섞을 횟수를 지정한다.(기본값 5)
)
print(perm_importance['importances_mean'])
````

`sugar`의 중요도가 가장 높다고 평가하였다.
>`permutation_importance()` 함수의 리턴값에는 특성 중요도(`importances`), 평균(`importances_mean`), 표준 편차(`importances_std`)

````python
perm_importance_test = permutation_importance(
    hgb, test_input, test_target, n_repeats=10, random_state=42, n_jobs=-1
)
print(perm_importance_test['importances_mean'])
````

테스트 세트도 비슷한 경향을 보인다.

마지막으로 `score()`을 사용하여 최종 성능을 확인한다.

````python
hgb.score(test_input, test_target)
````

앙상블 모델은 일반 결정 트리 모델보다 더욱 높은 점수를 얻을 수 있는 것을 확인하였다.
>히스토그램 기반 그레이디언트 부스팅의 회귀 모델은 `HistoGradientBoostingRegressor`클래스이다.

## 그레이디언트 부스팅 알고리즘들

그레이디언트 부스팅은 여러 개의 약한 학습기를 순차적으로 학습하고, 이전 단계의 예측 오차를 보완하도록 다음 학습기를 추가하는 앙상블 기법이다. 이 아이디어를 바탕으로 다양한 라이브러리와 개선 알고리즘이 개발되었다.

| 알고리즘 | 주요 특징 |
|---|---|
| `GradientBoostingClassifier` | scikit-learn의 기본 부스팅 모델이다. 트리를 순차적으로 추가하며, `learning_rate`, `n_estimators`, `max_depth` 등을 조정한다. |
| `HistGradientBoostingClassifier` | 특성 값을 구간(bin)으로 나누어 분할 후보를 효율적으로 탐색한다. scikit-learn에서 제공하며, 대용량 정형 데이터에서 기본 그레이디언트 부스팅보다 빠를 수 있다. |
| **XGBoost** | 정규화와 다양한 최적화 기법을 적용한 라이브러리다. 성능과 기능이 풍부하며, 매개변수를 세밀하게 조정할 수 있다. |
| **LightGBM** | 히스토그램 기반 학습을 사용해 빠르고 메모리 효율적인 편이다. 대규모 데이터에 유용하지만, 작은 데이터에서는 과적합에 주의해야 한다. |
| **CatBoost** | 범주형 특성을 다루는 기능이 특징이다. 범주형 데이터 전처리를 줄일 수 있지만, 사용 시 라이브러리의 입력 형식과 매개변수를 확인해야 한다. |

이들 알고리즘은 모두 부스팅을 기반으로 하지만, 트리 성장 방식, 범주형·결측값 처리, 정규화와 최적화 방법이 서로 다르다. 따라서 특정 알고리즘이 항상 가장 좋은 것은 아니며, 데이터의 크기와 특성, 학습 시간 및 검증 성능을 고려해 선택하는 것이 좋다. 성능 비교에는 동일한 데이터 분할이나 교차 검증을 사용하고, 최종 평가는 테스트 세트로 수행한다.

`XGBoost`를 이용해보겠다.
>`XGBoost`는 따로 설치해야 하는 새로운 라이브러리이다.

````python
from xgboost import XGBClassifier
xgb = XGBClassifier(tree_method='hist', random_state=42)
scores_xgb = cross_validate(xgb, train_input, train_target, return_train_score=True, n_jobs=-1)
print(np.mean(scores_xgb['train_score']), np.mean(scores_xgb['test_score']))
````

## LightGBM

**LightGBM**은 Microsoft에서 개발한 그레이디언트 부스팅 라이브러리로, 히스토그램 기반 분할을 사용해 대규모 정형 데이터를 빠르고 메모리 효율적으로 학습한다. scikit-learn과 비슷한 인터페이스를 제공하며, 분류에는 `LGBMClassifier`, 회귀에는 `LGBMRegressor`를 사용한다.

LightGBM의 특징 중 하나는 트리를 **리프 중심(leaf-wise)**으로 성장시키는 방식이다. 매 단계에서 손실을 가장 크게 줄이는 리프를 우선 분할하므로 학습이 빠르고 높은 성능을 낼 수 있다. 다만 데이터가 적거나 트리가 지나치게 복잡하면 과적합될 수 있어 `num_leaves`, `max_depth`, `min_child_samples` 등의 매개변수를 조정하는 것이 좋다.

- `n_estimators`: 부스팅 반복 횟수
- `learning_rate`: 각 트리의 기여도
- `num_leaves`: 트리 하나의 최대 리프 수
- `max_depth`: 트리의 최대 깊이
- `min_child_samples`: 리프에 필요한 최소 샘플 수

LightGBM은 범주형 특성도 지원하지만, 입력 데이터의 자료형과 범주형 특성 지정 방법을 확인해야 한다. 이 예제처럼 수치형 특성을 사용하는 경우에는 별도의 범주형 처리가 필요하지 않다. 다른 부스팅 모델과 마찬가지로 동일한 교차 검증 데이터와 기준으로 성능을 비교하는 것이 바람직하다.
>LightGBM을 사용하려면 환경에 따라 먼저 `lightgbm` 라이브러리를 설치해야 한다.

````python
from lightgbm import LGBMClassifier

lgbm = LGBMClassifier(random_state=42)
scores_lgbm = cross_validate(
    lgbm, train_input, train_target,
    return_train_score=True, n_jobs=-1
)
print(np.mean(scores_lgbm['train_score']), np.mean(scores_lgbm['test_score']))
````

`LightGBM`까지 사용해보았다.

### XGBoost와 LightGBM 비교

두 모델 모두 그래디언트 부스팅을 기반으로 하지만, 트리를 성장시키는 방식과 최적화 방법이 다르다.

- **XGBoost**는 정규화와 다양한 과적합 방지 기능을 제공하며, `tree_method='hist'`를 지정하면 히스토그램 기반으로 분할을 탐색할 수 있다.
- **LightGBM**은 히스토그램 기반 학습과 리프 중심(leaf-wise) 트리 성장을 사용한다. 학습이 빠르고 메모리 효율적인 편이지만, 데이터가 적거나 트리가 복잡하면 과적합에 주의해야 한다.

이 실험의 교차 검증 평균 점수는 LightGBM이 약 **0.880**, XGBoost가 약 **0.878**로 비슷하다. 다만 훈련 점수는 XGBoost가 더 높아, 이 설정에서는 훈련 데이터에 더 잘 맞지만 과적합 가능성도 살펴볼 필요가 있다. 이 결과만으로 어느 모델이 항상 우수하다고 결론 내릴 수는 없으며, 데이터와 매개변수에 따라 성능이 달라진다.

# 앙상블 학습 정리

## 1. 앙상블 학습이란?

**앙상블 학습(Ensemble Learning)**은 여러 학습 모델의 예측을 결합해 최종 예측을 만드는 기계학습 기법이다. 모델마다 서로 다른 오류를 보완할 수 있어 예측 성능과 안정성을 높이는 데 도움이 된다.

- **분류**: 다수결 투표 또는 클래스별 예측 확률 결합
- **회귀**: 예측값의 평균 또는 가중 평균
- **주의점**: 모델들이 비슷한 오류를 내거나 계산 비용이 커지면 앙상블의 이점이 줄어들 수 있다.

대표적인 앙상블 방식으로 **배깅(Bagging)**, **부스팅(Boosting)**, **스태킹(Stacking)**이 있다. 앙상블은 정형 데이터에서 널리 쓰이며, 비정형 데이터에는 신경망 기반 딥러닝이 자주 활용된다.

## 2. 랜덤 포레스트

**랜덤 포레스트(Random Forest)**는 여러 결정 트리를 독립적으로 학습시킨 뒤 예측을 결합하는 모델이다.

- 각 트리는 훈련 데이터에서 **복원 추출**한 부트스트랩 샘플로 학습한다.
- 각 노드에서는 일부 특성을 무작위로 선택해 분할 후보를 탐색한다.
- 분류는 다수결, 회귀는 트리 예측값의 평균으로 최종 결과를 정한다.
- 분류 모델의 기본 특성 선택 기준은 보통 전체 특성 수의 제곱근이다. 회귀 모델은 기본적으로 전체 특성을 고려한다.

랜덤 포레스트는 과적합을 줄이고 특성 중요도를 확인할 수 있지만, 단일 결정 트리보다 학습과 예측에 더 많은 자원을 사용하며 해석이 어렵다. `oob_score=True`를 지정하면 부트스트랩 샘플에 포함되지 않은 OOB 샘플을 이용해 모델을 평가할 수 있다.

## 3. 엑스트라 트리

**엑스트라 트리(Extra Trees)**는 여러 결정 트리를 결합한다는 점에서 랜덤 포레스트와 비슷하지만, 기본적으로 부트스트랩 샘플 대신 전체 훈련 데이터를 사용한다. 노드 분할 후보를 더 무작위로 선택해 학습 속도를 높일 수 있으며, `ExtraTreesClassifier`로 분류 문제에 적용한다.

## 4. 그레이디언트 부스팅

**그레이디언트 부스팅(Gradient Boosting)**은 얕은 결정 트리를 순차적으로 추가해 앞선 모델의 오차를 줄이는 방식이다. 랜덤 포레스트처럼 트리를 독립적으로 학습하지 않는다.

- `n_estimators`: 트리의 수
- `learning_rate`: 각 트리의 기여도
- `max_depth`: 개별 트리의 최대 깊이
- `subsample`: 각 단계에서 사용할 훈련 샘플의 비율

학습률과 트리 수는 함께 조정하는 것이 좋다. 트리를 순차적으로 학습하므로 병렬화가 제한되어 학습이 느릴 수 있다.

## 5. 히스토그램 기반 그레이디언트 부스팅

**히스토그램 기반 그레이디언트 부스팅**은 특성 값을 구간(bin)으로 나누어 분할 후보를 효율적으로 탐색한다. 일반적인 그레이디언트 부스팅보다 학습 속도와 메모리 효율이 개선될 수 있으며, 결측값 처리 기능도 제공한다.

scikit-learn에서는 `HistGradientBoostingClassifier`와 `HistGradientBoostingRegressor`를 제공한다. 주요 매개변수로는 `max_iter`, `learning_rate`, `max_depth`, `max_leaf_nodes`, `l2_regularization` 등이 있다. 이 모델은 `feature_importances_`를 제공하지 않으므로, 특성 중요도 분석에는 `permutation_importance()`를 사용할 수 있다.

## 6. 그레이디언트 부스팅 라이브러리

- **XGBoost**: 다양한 정규화와 최적화 기능을 제공한다. `tree_method='hist'`로 히스토그램 기반 학습을 사용할 수 있다.
- **LightGBM**: 히스토그램 기반 학습과 리프 중심 트리 성장을 사용한다. 빠르고 메모리 효율적일 수 있지만, 작은 데이터에서는 과적합에 주의해야 한다.
- **CatBoost**: 범주형 특성 처리 기능을 제공한다.

어떤 모델이 항상 가장 우수한 것은 아니다. 동일한 데이터 분할과 교차 검증 기준으로 비교하고, 검증 성능과 학습 시간 등을 함께 고려한다.

## 7. 와인 데이터셋 실험

와인 데이터셋에서 `alcohol`, `sugar`, `pH`를 입력 특성으로 사용하고 `class`를 예측한다. 데이터를 훈련 세트와 테스트 세트로 나눈 뒤, 다음 모델을 교차 검증으로 비교한다.

1. `RandomForestClassifier`
2. `ExtraTreesClassifier`
3. `GradientBoostingClassifier`
4. `HistGradientBoostingClassifier`
5. `XGBClassifier`
6. `LGBMClassifier`

`cross_validate()`를 사용해 훈련 점수와 검증 점수를 확인하고, 앙상블별 특성 중요도도 살펴본다. 실험에서는 여러 모델이 `sugar`를 중요한 특성으로 평가했다. 히스토그램 기반 부스팅의 특성 중요도는 훈련 데이터 또는 테스트 데이터에서 `permutation_importance()`로 확인할 수 있다.

마지막에는 테스트 세트의 `score()`로 최종 성능을 평가한다. 테스트 세트는 모델 선택이나 매개변수 조정에 반복적으로 사용하지 않고, 최종 평가에 활용하는 것이 좋다.

## 8. 핵심 정리

- 앙상블은 여러 모델의 예측을 결합해 성능과 안정성을 높이는 방법이다.
- 랜덤 포레스트와 엑스트라 트리는 여러 트리를 비교적 독립적으로 학습한다.
- 부스팅 모델은 트리를 순차적으로 추가해 이전 모델의 오류를 보완한다.
- 모델과 매개변수의 성능은 데이터에 따라 달라지므로 교차 검증으로 비교한다.
- 특성 중요도는 모델의 예측에 특성이 기여하는 정도를 살펴보는 참고 자료이며, 인과관계를 뜻하지는 않는다.
