````python
fish_target = [1] * 35 + [0] * 14
print(fish_target)
````

아래와 같이 `KNeighborsClassifier`를 불러옵니다.

또는 다음 코드와 같이 **객체의 인스턴스로 선언**하여 불러올 수 있습니다.
```python
import sklearn

```

````python
from sklearn.neighbors import KNeighborsClassifier
````

다음과 같이 클래스 인스턴스를 선언하고, `fit()` 메소드를 이용해 모델을 훈련시킬 수 있습니다.

아래에서는 `fish_data`와 `fish_target`을 매개변수로 입력하여 데이터를 입력합니다.
`KNeighborsClassifier`는 이를 토대로 모델을 훈련합니다.

### `fit()`
`fit()` 메소드의 첫 번째 매개변수에는 `fish_data`가 입력, 즉 훈련될 샘플 데이터 세트가 입력되고,
두 번째 매개변수에 입력된 `fish_target`은 훈련될 샘플 데이터 세트의 구분을 위한 확인 배열이 주어집니다.

이것을 `0`과 `1`로 구분하여, 구별할 수 있게 하였습니다.

````python
kn = KNeighborsClassifier()
kn.fit(fish_data, fish_target)
````

이제 모델의 정확도를 판단할 차례입니다.

다음과 같이 `score()` 메소드를 이용해 모델의 정확도를 판단할 수 있습니다.

### `score()`
`score()` 메소드는 모델에 대해 0에서 1의 자리 수 사이의 부동 소수점 소수 첫째 자리 값을 반환합니다.
이 값을 100배 하여 퍼센트로 나타내어 정확도를 평가할 수 있습니다.
> 예를 들어 `0.5`라면, `50%`를 맞추었다는, 즉 절반을 맞추었다는 것입니다.

아래에서는 당연히 훈련되었던 데이터 세트와 완전히 같은 데이터를 입력해 평가했으므로 `1.0`이 출력됩니다.

````python
kn.score(fish_data, fish_target)
````

# 데이터 세트 활용하기
이제 훈련되었고, 테스트(평가)된 데이터 세트를 이용해 실제 값을 구별할 차례입니다.

우선, 아래의 산점도를 **인간의 입장**에서 확인해봅시다.

````python
plt.scatter(bream_length, bream_weight)
plt.scatter(smelt_length, smelt_weight)
plt.scatter(30, 600, s=100, marker="^")
plt.xlabel('weigh') # x축 설명
plt.ylabel('length') # y축 설명
plt.show()
````

위의 산점도에서 삼각형으로 주어진 값이 파란색 마커 주변에 있으므로, 이것을 도미라 구분할 수 있을 것입니다.

삼각형 주변에 다른 **도미 데이터**가 많이 때문입니다.

knn도 마찬가지입니다.
주변의 가까운 데이터의 개수 중, 가장 많은 비율을 차지하는 구분값에 대해 값을 지정합니다.

실제로도 그런지, 학습된 모델을 사용하는 `predict()` 메소드를 이용해 확인해봅시다.
> 2차원 배열 형태로 데이터를 입력합니다.

````python
kn.predict([[30, 600]])
````

값은 `1`번, 즉 `bream`(도미)로 판정되었습니다.
이 삼각형 포인트는 도미가 맞았습니다.

이렇게 본다면, Knn은 그다지 복잡한 알고리즘이 아닙니다.
해당 데이터 주변의 가장 근접한 값들을 찾아내어, 그 값의 비율이 많은 것을 찾아내는 알고리즘이므로
**사실상 훈련되는 것은 없다고 봐도 무방합니다.**
그때그때 데이터를 분석하여 수학문제 풀듯이 풀이하면 되니까요.

### 단점
그래서 Knn은 많은 데이터를 계산하는 데에는 좋지 않습니다.
이를 위해 개선된 알고리즘(KD-트리, Ball-트리, 또는 근사 최근접 이웃 알고리즘 등)이 존재합니다.

### 내부
코드에서 확인해보면, 우리가 객체의 인스턴스로 생성한 `kn`에는 `_fit_X`와 `_y`를 통해 훈련되었던 값 확인이 가능합니다.

````python
print(kn._fit_X)
print(kn._y)
````

위와 같이 모든 값이 객체에 저장됩니다.

#### 참조값의 범위
위에서는 **데이터 주변의 가장 근접한 값들을 찾아내어,**라 하였습니다.
그렇다면, 그 **데이터 주변의 가장 근접한 값들**의 범위는 어떨까요?

그 값은 객체를 선언할 때, 기본 매개변수에 저장할 수 있습니다.

> `n_neighbors`의 기본값은 `5` 입니다.
> 5개의 이웃을 참조합니다.

````python
kn49 = KNeighborsClassifier(n_neighbors=49)
````

데이터 세트의 값 전체인 49개 모두를 참조하도록 했습니다.

모든 데이터를 참조하는데, 그 비율에서 도미가 빙어보다 많으므로 무조건 값은 `1`, 즉 도미로 판정되게 됩니다.

````python
kn49.fit(fish_data, fish_target)
kn49.score(fish_data, fish_target)
````

`0.7142857142857143`는 49개의 샘플 데이터 중, 도미의 개수 35개를 나눈 값입니다.

모든 데이터를 참조하는 것은 올바르지 않습니다.
데이터의 참조 개수를 적당히 조절하는 것이 중요합니다.
